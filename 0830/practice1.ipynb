```python
import random

min = 0
max = 100
tarset = random.randint(min,max)
count = 0

print("=====猜數字遊戲===== \n")
while(True):
    count = count+1
    keyin = int(input(f'輸入數字 {min} ~ {max}\n'))
    if (keyin == tarset):
        print(f"您猜了 {count} 次")
        print(f"賓果，答對了!! 答案是 {tarset}")
        break
    elif (keyin > tarset):
        print(f'您猜了 {count} 次')
        print(f'比 {keyin} 小')
        max = keyin
    elif (keyin < tarset):
        print(f'您猜了 {count} 次')
        print(f'比 {keyin} 大')
        min = keyin
print("遊戲結束")
```
